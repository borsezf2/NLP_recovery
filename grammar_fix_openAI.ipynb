```python
from openai import OpenAI
import time
import os
from ratelimit import limits, sleep_and_retry


# Set your OpenAI API key as an environment variable
os.environ["OPENAI_API_KEY"] = ""
```

```python
# Define rate limit: 3 calls per minute
CALLS = 3
RATE_LIMIT = 60

@limits(calls=CALLS, period=RATE_LIMIT)
# def generate_text(prompt, max_tokens=400, model="gpt-3.5-turbo"):
def generate_text(prompt, max_tokens=400, model="gpt-4o"):
    client = OpenAI()
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": "You are a grammar fixing tool, you'll be given a paragraph to fix the grammar and list out all the changed you made."},
            {"role": "user", "content": prompt}
        ],
        max_tokens=max_tokens,
    )
    return response.choices[0].message.content.strip()
```

```python

# Define your prompt

# Ask a coding question


prompt = ''' 
Thomas is a student on IIT Kharagpur. Where he is studying datascience and he always dreamed to becomes a machine learning enginer. 
Thoms is a proactived student, and and always looked for research related opportunity.
'''

try:
    for i in range(1):  # Try to make 1 calls
        result = generate_text(prompt)
        print(f"Call {i+1}: {result}")
except Exception as e:
    print(f"An error occurred: {e}")
```
